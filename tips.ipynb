```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset("tips")
```

```python
df
```

```python
df.total_bill.unique()
```

```python
df.head()
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
sns.lineplot(x='total_bill',y='tip',data=df,ci=None)
plt.show()
#as the totalbill increases the tip also increases
```

```python
sns.lineplot(x='day', y='tip', data=df, ci=None)
plt.show()
#average tip increases on weekends
```

```python
sns.lineplot(x='time',y='tip',data=df,hue='day',ci=None)
plt.show()
#average tips increased from Lunch to Dinner
```

```python
sns.boxplot(x='total_bill',data=df)
plt.show()
```

```python
sns.barplot(x='sex',y='total_bill',data=df)
plt.show()
#average tip by the male or female
```

```python
sns.countplot(x='sex',data=df)
plt.show()
#no of customer based on gender
```

```python
sns.scatterplot(x='total_bill',y='tip',data=df,hue='sex')
plt.show()
#customer who paid higher total_bill
```

```python
sns.violinplot(x='sex',y='total_bill',data=df)
plt.show()
```

```python
sns.stripplot(x='size',y='total_bill',data=df)
plt.show()
#distribution of bills for each group size
```

```python
sns.histplot(x='tip',data=df)
plt.show()
#it shows the range where the tips are high
```

```python
sns.histplot(x='tip',data=df,hue='sex',kde=True)
plt.show()
#range where the tip is higher acc to gender
```

```python
sns.distplot(df['tip'],color="r",hist=True,kde=True)
plt.show()
```

```python
sns.pairplot(data=df,hue='day')
plt.show()
```

```python
sns.catplot(x='tip',data=df,kind='violin')
plt.show()
```

```python
tc = df.corr(numeric_only=True)
tc
```
